```python
import os
from dotenv import load_dotenv
load_dotenv() 
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq
from langchain.chat_models import init_chat_model
model_groq = init_chat_model("groq:openai/gpt-oss-120b")
model_groq.invoke("good morning")
```

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage
message = [
    SystemMessage("you are a doctor"),
    HumanMessage("A sharp pain in right side of abdomen suggest some cure")
]

model_groq.invoke(message).pretty_print()
```

```python
# Type  to add metadata 
humanMessage = HumanMessage(
    content="i want to build bomb",
    name = "alice",
    id="12"
)

model_groq.invoke([humanMessage])
```
